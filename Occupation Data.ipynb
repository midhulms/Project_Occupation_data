### Q. Add a new column for age group (<30, 30-50, >50) and display the first 5 rows
(Hint: You can create a function for each age group)

```python
def age_group(age):
    if age < 30:
        return '<30'
    elif age <= 50:
        return '30-50'
    else:
        return '>50'
```

```python
users['age_group'] = users['age'].apply(age_group)
users.head()
```

### Q. Count the number of users in each age group by gender.

```python
user_age=users.groupby(['age_group', 'gender'])['gender'].count()
```

```python
user_age
```

### Q. Shows how the ages of users are distributed across different intervals
(Hint: Use a histogram chart)

```python
plt.figure(figsize=(8, 5))
plt.hist(users['age'], bins=10, color='steelblue', edgecolor='black')
plt.title('Age Distribution of Users')
plt.xlabel('Age')
plt.ylabel('Number of Users')
plt.show()
```

### Q. Shows the proportion of categories. Here, we can see how many users are male vs female.
(Hint: Use a Pie Chart)

```python
gender_counts = users['gender'].value_counts()
```

```python
plt.figure(figsize=(6, 6))
plt.pie(gender_counts, labels=['Male', 'Female'], autopct='%1.1f%%', colors=['steelblue', 'pink'])
plt.title('Male vs Female Users')
plt.show()
```
